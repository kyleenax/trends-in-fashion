```python
import subprocess
import sys

def install(package):
    subprocess.check_call([sys.executable, "-m", "pip", "install", package])
    
install("yfinance")
import yfinance as yf
sp500 = yf.Ticker("^GSPC")
sp500 = sp500.history(period="max")
sp500.index
```

```python
sp500.plot.line(y="Close",use_index=True)
del sp500["Dividends"]
del sp500["Stock Splits"]
```

```python
sp500["Tomorrow"] = sp500["Close"].shift(-1)
sp500["Target"] = (sp500["Tomorrow"] > sp500["Close"]).astype(int)
sp500 = sp500.loc["1990-01-01":].copy()
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, min_samples_split=100,random_state=1)

train=sp500.iloc[:-100]
test=sp500[-100:]

predictors=["Close","Volume","Open","High","Low"]
model.fit(train[predictors],train["Target"])
```

```python
from sklearn.metrics import precision_score
import pandas as pd
pred = model.predict(test[predictors])
pred = pd.Series(pred, index=test.index)
precision_score(test["Target"], pred)
```

```python
combined = pd.concat([test["Target"], pred], axis=1)
combined.plot()
```

```python
def predict(train, test, predictors, model):
    model.fit(train[predictors], train["Target"])
    preds = model.predict(test[predictors])
    preds = pd.Series(preds, index=test.index, name="Predictions")
    combined = pd.concat([test["Target"], preds], axis=1)
    return combined
```

```python

def backtest(data, model, predictors, start=2500, step=250):
    all_predictions = []

    for i in range(start, data.shape[0], step):
        train = data.iloc[0:i].copy()
        test = data.iloc[i:(i+step)].copy()
        predictions = predict(train, test, predictors, model)
        all_predictions.append(predictions)
    
    return pd.concat(all_predictions)
```

```python
predictions = backtest(sp500, model, predictors)
```

```python
from numpy import asarray
from pandas import read_csv
from pandas import DataFrame
from pandas import concat
from sklearn.metrics import mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
from matplotlib import pyplot
import yfinance as yf
import numpy as np
from datetime import datetime, timedelta
from sklearn.linear_model import LinearRegression



def get_dates_after(start_date_str, num_days):
    """
    Returns a list of dates after a given start date.

    Args:
        start_date_str (str): Start date in 'YYYY-MM-DD' format.
        num_days (int): Number of days after the start date.

    Returns:
        list: List of dates (strings in 'YYYY-MM-DD' format).
    """
    start_date = datetime.strptime(start_date_str, '%Y-%m-%d').date()
    dates_after = []
    for i in range(1, num_days + 1):
        next_date = start_date + timedelta(days=i)
        dates_after.append(next_date.strftime('%Y-%m-%d'))
    return dates_after

def series_to_supervised(data, n_in=1, n_out=1, dropnan=True):
    # If the data is a pandas Series, convert it to a 2D array with one column
    if isinstance(data, DataFrame):
        n_vars = data.shape[1]
    else:
        data = DataFrame(data)  # Convert the Series to DataFrame
        n_vars = 1  # Now data has one column
    cols = list()
    # input sequence (t-n, ... t-1)
    for i in range(n_in, 0, -1):
        cols.append(data.shift(i))
    # forecast sequence (t, t+1, ... t+n)
    for i in range(0, n_out):
        cols.append(data.shift(-i))
    # put it all together
    agg = concat(cols, axis=1)
    # drop rows with NaN values
    if dropnan:
        agg.dropna(inplace=True)
    return agg.values

# split a univariate dataset into train/test sets
def train_test_split(data, n_test):
    return data[:-n_test, :], data[-n_test:, :]

# fit a random forest model and make a one step prediction
def random_forest_forecast(train, testX):
    # Ensure train is a NumPy array
    train = np.array(train)
    # split into input and output columns
    trainX, trainy = train[:, :-1], train[:, -1]
    # fit model
    model = RandomForestRegressor(n_estimators=1000, warm_start=True)
    model.fit(trainX, trainy)
    # make a one-step prediction
    testX = asarray(testX).reshape(1, -1)  # Ensure testX is 2D
    yhat = model.predict(testX)
    return yhat[0]

def linear_regression_forecast(train, testX):
    # Ensure train is a NumPy array
    train = np.array(train)
    # split into input and output columns
    trainX, trainy = train[:, :-1], train[:, -1]
    # fit model
    model = LinearRegression()
    model.fit(trainX, trainy)
    # make a one-step prediction
    testX = asarray(testX).reshape(1, -1)  # Ensure testX is 2D
    yhat = model.predict(testX)
    return yhat[0]

# walk-forward validation for univariate data
def walk_forward_validation(data, n_test, dates):
    rf_predictions = list()  # List for Random Forest predictions
    lr_predictions = list()  # List for Linear Regression predictions
    y_true = list()  # List to store actual values
    # split dataset
    data = np.array(data)
    train, test = train_test_split(data, n_test)
    
    # Seed history with training dataset
    history = [x for x in train]
    
    # Ensure we get the previous `n_test` dates before the last date
    test_dates = dates[-n_test:] 
    # step over each time-step in the test set
    for i in range(len(test)):
        # split test row into input and output columns
        testX, testy = test[i, :-1], test[i, -1]
        
        # Random Forest prediction
        rf_yhat = random_forest_forecast(history, testX)
        rf_predictions.append(rf_yhat)
        
        # Linear Regression prediction
        lr_yhat = linear_regression_forecast(history, testX)
        lr_predictions.append(lr_yhat)
        
        y_true.append(testy)  # Store the true value
        
        # add actual observation to history for the next loop
        history.append(test[i])
        # summarize progress
        print('>expected=%.1f, RF predicted=%.1f, LR predicted=%.1f' % (testy, rf_yhat, lr_yhat))
    
    # Convert predictions into NumPy arrays of floats
    rf_predictions = np.array(rf_predictions, dtype=float)
    lr_predictions = np.array(lr_predictions, dtype=float)
    
    # Calculate the Mean Absolute Percentage Error (MAPE) for both models
    rf_mape = mean_absolute_percentage_error(y_true, rf_predictions)
    lr_mape = mean_absolute_percentage_error(y_true, lr_predictions)
    
    # Estimate prediction error for both models
    rf_error = mean_absolute_error(y_true, rf_predictions)
    lr_error = mean_absolute_error(y_true, lr_predictions)
    
    return rf_error, lr_error, rf_mape, lr_mape, y_true, rf_predictions, lr_predictions, test_dates


def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

def predict_future_values(data, n_forward, dates):
    predictions_rf = list()
    predictions_lr = list()
    
    # Use the last date from the given data as the base
    last_date = dates[-1]  # This is the last date of the test set
    print(last_date)
    future_dates = get_dates_after(last_date.strftime('%Y-%m-%d'), n_forward)
    
    # Convert data to a NumPy array for processing
    data = np.array(data)
    
    # Seed history with the training dataset
    history = data.copy()  # Create a copy of the data to avoid modifying the original

    window_size = 5
    
    for i in range(n_forward):
        history = np.array(history)
    
        # Ensure we have enough data
        if len(history) < window_size:
            testX = history[-1, :-1]  # Fallback
        else:
            testX = np.mean(history[-window_size:, :-1], axis=0)  # Use multiple past rows
    
        testX = testX.reshape(1, -1)  # Ensure correct shape
    
        rf_yhat = random_forest_forecast(history, testX)
        predictions_rf.append(rf_yhat)
    
        lr_yhat = linear_regression_forecast(history, testX)
        predictions_lr.append(lr_yhat)
    
        new_row = np.append(testX[-(history.shape[1] - 1):], rf_yhat)
        history = np.vstack([history, new_row])
    
        print(f'>RF Prediction={rf_yhat}, LR Prediction {lr_yhat}')


    
    # Convert predictions into NumPy arrays of floats
    predictions_rf = np.array(predictions_rf, dtype=float)
    predictions_lr = np.array(predictions_lr, dtype=float)
    
    # Return the future dates and both sets of predictions
    return future_dates, predictions_rf, predictions_lr

       

# Download historical S&P 500 data
sp500 = yf.Ticker("^GSPC")
sp500 = sp500.history(period="max")
sp500.index

# Remove Dividends and Stock Splits columns
del sp500["Dividends"]
del sp500["Stock Splits"]
sp500 = sp500.iloc[-2000:]

# Prepare the data for supervised learning
data = series_to_supervised(sp500["Close"])

# Walk-forward validation for forecasting
rf_mae, lr_mae, rf_mape, lr_mape, y, rf_yhat, lr_yhat, test_dates = walk_forward_validation(data, 50, sp500.index)
# Print the results
print(f'Random Forest Mean Absolute Error (MAE): {rf_mae}')
print(f'Linear Regression Mean Absolute Error (MAE): {lr_mae}')
print(f'Random Forest Mean Absolute Percentage Error (MAPE): {rf_mape}%')
print(f'Linear Regression Mean Absolute Percentage Error (MAPE): {lr_mape}%')
# Predict future values for the next 20 days using both models
future_dates, rf_future_predictions, lr_future_predictions = predict_future_values(data, 20, sp500.index)
print("done")

# # Print future predictions
# print("Future predictions:", future_predictions)

# # Plot future predictions
# pyplot.legend()
# pyplot.show()
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
from pandas import DataFrame, concat
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

def get_season_after(start_date_str, num_seasons, skip_off):
    """
    Returns a list of dates after a given start date, simulating seasons.
    """
    try:
        start_date = datetime.strptime(start_date_str, '%m/%d/%y').date()
    except ValueError:
        start_date = datetime.strptime(start_date_str, '%m/%d/%Y').date()

    dates_after = []
    if skip_off:
        for i in range(1, num_seasons + 1):
            next_date = start_date + timedelta(days=i)
            dates_after.append(next_date.strftime('%m/%d/%y'))
    else:
        for i in range(1, num_seasons + 1):
            next_date = start_date + timedelta(days=int((365/2) * i))
            dates_after.append(next_date.strftime('%m/%d/%y'))
    return dates_after

def random_forest_forecast(history, testX):
    model = RandomForestRegressor(n_estimators=100, random_state=1)
    model.fit(history[:-1], history[1:].ravel())  # Use ravel() here
    yhat = model.predict(testX)
    return yhat[0]

def linear_regression_forecast(history, testX):
    model = LinearRegression()
    model.fit(history[:-1], history[1:].ravel())  # Use ravel() here
    yhat = model.predict(testX)
    return yhat[0]

def predict_future_values(data, n_forward, dates):
    predictions_rf = list()
    predictions_lr = list()
    last_date = dates[-1]
    future_dates = get_season_after(last_date, n_forward, False)
    data = np.array(data)
    history = data.copy().reshape(-1, 1)
    window_size = 5
    for i in range(n_forward):
        history = np.array(history)
        if len(history) < window_size:
            testX = history[-1].reshape(1, -1)
        else:
            testX = np.mean(history[-window_size:], axis=0).reshape(1, -1)
        rf_yhat = random_forest_forecast(history, testX)
        predictions_rf.append(rf_yhat)
        lr_yhat = linear_regression_forecast(history, testX)
        predictions_lr.append(lr_yhat)
        new_row = np.array([[rf_yhat]])
        history = np.vstack([history, new_row])
        print(f'>RF Prediction={rf_yhat}, LR Prediction {lr_yhat}')
    return future_dates, predictions_rf, predictions_lr

```
